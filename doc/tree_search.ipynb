# Searching the Topic Tree

`TopicTreeSearch` provides a small `ipywidgets` search control around Toponymy's existing expandable HTML topic tree. It supports keyword and semantic search, annotates each topic with the matching-document count, and can expand leaf topics to show matching document snippets.

````python
from types import SimpleNamespace

import numpy as np
from IPython.display import display

from toponymy.plotting import TopicTreeSearch

documents = [
    'Graph neural networks learn representations from connected data.',
    'Community detection identifies groups in social networks.',
    'Telescope observations measure distant galaxies.',
    'Galaxy surveys map the large-scale structure of the universe.',
]

# This compact fixture has document vectors for tree construction only. It does
# not have an embedding model, so its explorer intentionally supports keywords.
model = SimpleNamespace(
    embedding_vectors_=np.asarray(
        [[1.0, 0.0], [0.8, 0.2], [0.0, 1.0], [0.1, 0.9]]
    ),
    cluster_tree_={(2, 0): [(1, 0), (1, 1)], (1, 0): [(0, 0)], (1, 1): [(0, 1)]},
    cluster_layers_=[
        SimpleNamespace(cluster_labels=np.asarray([0, 0, 1, 1])),
        SimpleNamespace(cluster_labels=np.asarray([0, 0, 1, 1])),
    ],
    topic_names_=[['Network methods', 'Astronomy'], ['Computing', 'Science']],
    topic_sizes_=[[2, 2], [2, 2]],
)

search = TopicTreeSearch(model, documents)
````

## Explore the tree

This compact, self-contained example supports keyword search. Search matches are counted at every topic in the native expandable tree. Select **Topic names** to search generated labels instead of source documents, and select **Show matching documents** to expand matching snippets beneath their most detailed topic.

````python
widget = search.explorer()
widget
````
